```python

file_name = list(uploaded.keys())[0] #extract the name of your file
with open(file_name, 'r', encoding='utf-8') as f: #open the text in read mode
    text = f.read()


```

```python

# Remove multiple newlines and replace them with a single space
import re
text = re.sub(r'\n+', ' ', text)  # collapse multiple newlines into one space
text = re.sub(r'\s+', ' ', text).strip()  # normalize whitespace
```

With the following cell, the NLP pipeline included in the SpaCy model is applied to your text.

```python
# STEP 4: Process the text
doc = nlp(text)
```

```python
text
```

By running the following cell, the different layers of linguistic annotation are extracted and saved in a **conllu** file.

The layers are:
- lemma
- part-of-speech
- morphological features
- dependency relations

```python
conllu_lines = []
for i, token in enumerate(doc): #extracting the annotation for each token of the text
    ID = i + 1
    FORM = token.text
    LEMMA = token.lemma_
    UPOS = token.pos_
    XPOS = token.tag_
    FEATS = token.morph if token.morph else "_"
    HEAD = token.head.i + 1 if token.head != token else 0
    DEPREL = token.dep_
    DEPS = "_"
    MISC = "_"
    conllu_lines.append(f"{ID}\t{FORM}\t{LEMMA}\t{UPOS}\t{XPOS}\t{FEATS}\t{HEAD}\t{DEPREL}\t{DEPS}\t{MISC}") #appending the annotation to our list

# Save to file
with open("syntactic_analysis.conllu", "w", encoding="utf-8") as f:
    f.write("\n".join(conllu_lines))

print("Saved as syntactic_analysis.conllu")
```
