Import the python library *Spacy* and load the downloaded model

```python
import spacy
nlp = spacy.load("la_core_web_sm")
```

Use the following two cells to import the text on which you want to perform the annotation. The text must be in **.txt** format.

```python

from google.colab import files
```

```python
uploaded = files.upload()
```

```python
#the text is saved in the variable 'text'
file_name = list(uploaded.keys())[0]
with open(file_name, 'r', encoding='utf-8') as f:
    text = f.read()

```

With the following cell, the NLP pipeline included in the LatinCy model is applied to your text.

```python
doc = nlp(text)
```

By running the following cell, the different layers of linguistic annotation are extracted and saved in a **conllu** file.

The layers are:
- lemma
- part-of-speech
- morphological features
- dependency relations

```python

conllu_lines = []
for i, token in enumerate(doc):
    ID = i + 1
    FORM = token.text
    LEMMA = token.lemma_
    UPOS = token.pos_
    XPOS = token.tag_
    FEATS = token.morph if token.morph else "_"
    HEAD = token.head.i + 1 if token.head != token else 0
    DEPREL = token.dep_
    DEPS = "_"
    MISC = "_"
    conllu_lines.append(f"{ID}\t{FORM}\t{LEMMA}\t{UPOS}\t{XPOS}\t{FEATS}\t{HEAD}\t{DEPREL}\t{DEPS}\t{MISC}")

# Save to file
with open("latin_analysis.conllu", "w", encoding="utf-8") as f:
    f.write("\n".join(conllu_lines))

print("Saved as latin_analysis.conllu")


```
