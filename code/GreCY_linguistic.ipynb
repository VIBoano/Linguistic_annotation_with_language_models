Perform the necessary imports and load the downloaded model, then upload your file. The file must be in **.txt** format.




```python
import spacy
from google.colab import files
from spacy import displacy
```

```python
nlp = spacy.load("grc_odycy_joint_sm") #loading the model
```

```python
uploaded = files.upload() #upload your file
```

```python
file_name = list(uploaded.keys())[0] #extract the name of the file
with open(file_name, 'r', encoding='utf-8') as f: #open the file in reading mode
    text = f.read()
```

With the following cell, the NLP pipeline included in the OdyCy model is applied to your text.

```python
doc = nlp(text) #process the text with the model
```

By running the following cell, the different layers of linguistic annotation are extracted and saved in a **conllu** file.

The layers are:
- lemma
- part-of-speech
- morphological features
- dependency relations

```python
for token in doc:
   print(token.orth_, token.lemma_,token.is_stop, token.pos_, token.morph, token.dep_, token.head)
```

```python
conllu_lines = []
for i, token in enumerate(doc):  #extracting the annotation for each token of the text
    ID = i + 1
    FORM = token.text
    LEMMA = token.lemma_
    UPOS = token.pos_
    XPOS = token.tag_
    FEATS = token.morph if token.morph else "_"
    HEAD = token.head.i + 1 if token.head != token else 0
    DEPREL = token.dep_
    DEPS = "_"
    MISC = "_"
    conllu_lines.append(f"{ID}\t{FORM}\t{LEMMA}\t{UPOS}\t{XPOS}\t{FEATS}\t{HEAD}\t{DEPREL}\t{DEPS}\t{MISC}")

# Save to file
with open("syntactic_analysis_greek.conllu", "w", encoding="utf-8") as f:
    f.write("\n".join(conllu_lines))

print("Saved as syntactic_analysis_greek.conllu")
```

The last cell displays the annotation with displacy

```python
displacy.serve(doc)
```
